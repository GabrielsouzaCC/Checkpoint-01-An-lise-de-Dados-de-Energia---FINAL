```python
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
print('Bibliotecas carregadas.')
```

1 A

```python
# 1. Carregar amostra
ds1 = pd.read_csv('amostra_ds1.csv')
print(ds1.head())
print('Shape:', ds1.shape)
ds1.info()
print(ds1.describe())
```

```python
# 2. Renomear atributos
ds1 = ds1.rename(columns={
    'Appliances': 'Consumo_Eletrodomesticos',
    'T1': 'Temp_Sala',
    'T2': 'Temp_Cozinha',
    'T3': 'Temp_Quarto',
    'RH_1': 'Umidade_Sala',
    'RH_2': 'Umidade_Cozinha',
    'RH_3': 'Umidade_Quarto'
})
print(ds1.columns.tolist())
```

```python
# 3. Maior consumo
max_c = ds1['Consumo_Eletrodomesticos'].max()
print(f'Maior consumo: {max_c} Wh')
```

```python
# 4. Limiar 70% — alta demanda
limiar = 0.70 * max_c
alta = ds1[ds1['Consumo_Eletrodomesticos'] > limiar]
qtd  = len(alta)
pct  = (qtd / len(ds1)) * 100
print(f'Limiar (70%): {limiar:.2f} Wh')
print(f'Registros acima: {qtd} ({pct:.2f}%)')
print(alta.head())
```

```python
# 5. Segundo critério: alto consumo E temperatura acima da média
temp_media = ds1['Temp_Sala'].mean()
alta_e_quente = ds1[
    (ds1['Consumo_Eletrodomesticos'] > limiar) &
    (ds1['Temp_Sala'] > temp_media)
]
qtd2 = len(alta_e_quente)
pct2 = (qtd2 / len(ds1)) * 100
print(f'Temperatura média: {temp_media:.2f} °C')
print(f'Alto consumo + temp alta: {qtd2} registros ({pct2:.2f}%)')
```

```python
# 6. Comparação
print(f'Apenas alto consumo:         {qtd} registros ({pct:.2f}%)')
print(f'Alto consumo + temp alta:    {qtd2} registros ({pct2:.2f}%)')
print(f'Redução: {qtd - qtd2} registros ao adicionar a condição de temperatura.')
```

```python
# Gráfico — Distribuição do consumo
plt.figure(figsize=(8, 4))
plt.hist(ds1['Consumo_Eletrodomesticos'], bins=30, color='steelblue', edgecolor='white')
plt.axvline(limiar, color='red', linestyle='--', label=f'Limiar 70% ({limiar:.0f} Wh)')
plt.title('Dataset 1 — Distribuição do Consumo de Eletrodomésticos')
plt.xlabel('Consumo (Wh)')
plt.ylabel('Frequência')
plt.legend()
plt.tight_layout()
plt.savefig('ds1_grafico.png', dpi=100)
plt.close()
print('Gráfico salvo: ds1_grafico.png')
```

```python
# 1. Carregar amostra
ds2 = pd.read_csv('amostra_ds2.csv')
print(ds2.head())
print('Shape:', ds2.shape)
ds2.info()
print(ds2.describe())
```

```python
# 2. Renomear atributos
ds2 = ds2.rename(columns={
    'Usage_kWh':                    'Consumo_kWh',
    'Leading_Current_Power_Factor':  'FP_Corrente',
    'Lagging_Current_Power_Factor':  'FP_Atraso'
})
print(ds2.columns.tolist())
```

```python
# 3. Maior consumo e limiar 75%
max_c2  = ds2['Consumo_kWh'].max()
limiar2 = 0.75 * max_c2
print(f'Maior consumo: {max_c2:.2f} kWh')
print(f'Limiar (75%):  {limiar2:.2f} kWh')
```

```python
# 4. DataFrame de alta demanda
alta2 = ds2[ds2['Consumo_kWh'] > limiar2]
qtd2a  = len(alta2)
pct2a  = (qtd2a / len(ds2)) * 100
print(f'Registros acima do limiar: {qtd2a} ({pct2a:.2f}%)')
```

```python
# 5. Quantos pertencem a Maximum Load?
if 'Load_Type' in ds2.columns:
    max_load = alta2[alta2['Load_Type'] == 'Maximum_Load']
    print(f'Registros Maximum_Load na alta demanda: {len(max_load)}')
else:
    print('Coluna Load_Type não encontrada — verifique o nome no seu arquivo.')
```

```python
# 6. Segundo critério: consumo elevado E fator de potência baixo (< 0.85)
fp_limite = 0.85
alta2_fp  = ds2[
    (ds2['Consumo_kWh'] > limiar2) &
    (ds2['FP_Atraso'] < fp_limite)
]
print(f'Consumo alto + FP < {fp_limite}: {len(alta2_fp)} registros')
print(alta2_fp[['Consumo_kWh', 'FP_Atraso']].head())
```

```python
# Gráfico — Consumo vs Fator de Potência
plt.figure(figsize=(7, 4))
plt.scatter(ds2['FP_Atraso'], ds2['Consumo_kWh'], alpha=0.4, color='steelblue', s=10)
plt.axhline(limiar2,    color='red',    linestyle='--', label=f'Limiar ({limiar2:.1f} kWh)')
plt.axvline(fp_limite,  color='orange', linestyle='--', label=f'FP limite ({fp_limite})')
plt.title('Dataset 2 — Consumo vs Fator de Potência')
plt.xlabel('Fator de Potência (Atraso)')
plt.ylabel('Consumo (kWh)')
plt.legend()
plt.tight_layout()
plt.savefig('ds2_grafico.png', dpi=100)
plt.close()
print('Gráfico salvo: ds2_grafico.png')
```

```python
# 1. Carregar amostra
ds3 = pd.read_csv('amostra_ds3.csv')
print(ds3.head())
print('Shape:', ds3.shape)
ds3.info()
print(ds3.describe())
```

```python
# 2. Renomear colunas de consumo
# Ajuste os nomes do lado esquerdo conforme aparecerem no seu arquivo
ds3 = ds3.rename(columns={
    'Zone 1 Power Consumption': 'Consumo_Zona_1',
    'Zone 2  Power Consumption': 'Consumo_Zona_2',
    'Zone 3  Power Consumption': 'Consumo_Zona_3'
})
print(ds3.columns.tolist())
```

```python
# 3. Máximo por zona
max_z1 = ds3['Consumo_Zona_1'].max()
max_z2 = ds3['Consumo_Zona_2'].max()
max_z3 = ds3['Consumo_Zona_3'].max()
print(f'Zona 1 max: {max_z1:.2f}')
print(f'Zona 2 max: {max_z2:.2f}')
print(f'Zona 3 max: {max_z3:.2f}')
```

```python
# 4. Zona com maior pico
maximos = {'Consumo_Zona_1': max_z1, 'Consumo_Zona_2': max_z2, 'Consumo_Zona_3': max_z3}
zona_pico = max(maximos, key=maximos.get)
valor_pico = maximos[zona_pico]
print(f'Zona com maior pico: {zona_pico} ({valor_pico:.2f})')
```

```python
# 5. Limiar 70% da zona de maior pico
limiar3 = 0.70 * valor_pico
alta3   = ds3[ds3[zona_pico] > limiar3]
qtd3    = len(alta3)
pct3    = (qtd3 / len(ds3)) * 100
print(f'Limiar: {limiar3:.2f}')
print(f'Registros acima: {qtd3} ({pct3:.2f}%)')
```

```python
# 6. Segundo critério: consumo acima do limiar E temperatura acima da média
temp_media3  = ds3['Temperature'].mean()
alta3_temp   = ds3[
    (ds3[zona_pico] > limiar3) &
    (ds3['Temperature'] > temp_media3)
]
qtd3b = len(alta3_temp)
pct3b = (qtd3b / len(ds3)) * 100
print(f'Temperatura média: {temp_media3:.2f} °C')
print(f'Alto consumo + temp alta: {qtd3b} registros ({pct3b:.2f}%)')
print(f'Redução: {qtd3 - qtd3b} registros ao adicionar temperatura.')
```

```python
# Gráfico — Comparação de consumo entre zonas
plt.figure(figsize=(8, 4))
plt.plot(ds3['Consumo_Zona_1'].values, label='Zona 1', alpha=0.7)
plt.plot(ds3['Consumo_Zona_2'].values, label='Zona 2', alpha=0.7)
plt.plot(ds3['Consumo_Zona_3'].values, label='Zona 3', alpha=0.7)
plt.title('Dataset 3 — Consumo por Zona (amostra)')
plt.xlabel('Índice do registro')
plt.ylabel('Consumo (kW)')
plt.legend()
plt.tight_layout()
plt.savefig('ds3_grafico.png', dpi=100)
plt.close()
print('Gráfico salvo: ds3_grafico.png')
```

```python
# 1. Carregar amostra
ds5 = pd.read_csv('amostra_ds5.csv')
print(ds5.head())
print('Shape:', ds5.shape)
ds5.info()
print(ds5.describe())
```

```python
# 2. Identificar e renomear colunas de produção
# Ajuste os nomes conforme aparecerem no seu arquivo
print('Colunas disponíveis:', ds5.columns.tolist())
```

```python
# Após ver as colunas acima, ajuste o rename abaixo:
ds5 = ds5.rename(columns={
    ds5.columns[1]: 'Producao_Eolica',  # segunda coluna — ajuste se necessário
    ds5.columns[2]: 'Producao_Solar'    # terceira coluna — ajuste se necessário
})
print(ds5[['Producao_Eolica', 'Producao_Solar']].describe())
```

```python
# 3. Máximos e limiares
max_eol  = ds5['Producao_Eolica'].max()
max_sol  = ds5['Producao_Solar'].max()
lim_eol  = 0.80 * max_eol
lim_sol  = 0.80 * max_sol
print(f'Eólica — max: {max_eol:.2f} | limiar: {lim_eol:.2f}')
print(f'Solar  — max: {max_sol:.2f} | limiar: {lim_sol:.2f}')
```

```python
# 4. Períodos de alta produção em cada fonte
alta_eol = ds5[ds5['Producao_Eolica'] > lim_eol]
alta_sol = ds5[ds5['Producao_Solar']  > lim_sol]
print(f'Alta eólica: {len(alta_eol)} registros ({len(alta_eol)/len(ds5)*100:.2f}%)')
print(f'Alta solar:  {len(alta_sol)} registros ({len(alta_sol)/len(ds5)*100:.2f}%)')
```

```python
# 5. Segundo critério: alta produção simultânea (eólica E solar)
alta_ambas = ds5[
    (ds5['Producao_Eolica'] > lim_eol) &
    (ds5['Producao_Solar']  > lim_sol)
]
print(f'Alta produção simultânea: {len(alta_ambas)} registros ({len(alta_ambas)/len(ds5)*100:.2f}%)')
```

```python
# Gráfico — Produção eólica vs solar
plt.figure(figsize=(8, 4))
plt.plot(ds5['Producao_Eolica'].values, label='Eólica', color='steelblue', alpha=0.7)
plt.plot(ds5['Producao_Solar'].values,  label='Solar',  color='orange',    alpha=0.7)
plt.title('Dataset 5 — Produção Eólica e Solar (amostra)')
plt.xlabel('Índice do registro')
plt.ylabel('Produção (MWh)')
plt.legend()
plt.tight_layout()
plt.savefig('ds5_grafico.png', dpi=100)
plt.close()
print('Gráfico salvo: ds5_grafico.png')
```

```python
# 1. Carregar amostra
# Se vier direto do arquivo original (sem Orange), use:
# ds6 = pd.read_csv('household_power_consumption.txt', sep=';', na_values='?')
# Com a amostra exportada do Orange:
ds6 = pd.read_csv('amostra_ds6.csv')
print(ds6.head())
print('Shape:', ds6.shape)
ds6.info()
print(ds6.describe())
```

```python
# 2. Renomear atributos
ds6 = ds6.rename(columns={
    'Global_active_power':   'Potencia_Ativa',
    'Global_reactive_power': 'Potencia_Reativa',
    'Sub_metering_1':        'Cozinha',
    'Sub_metering_2':        'Lavanderia',
    'Sub_metering_3':        'Climatizacao'
})
print(ds6.columns.tolist())
```

```python
# 3. Verificar e remover valores ausentes
print('Valores ausentes antes:')
print(ds6.isnull().sum())
ds6 = ds6.dropna(subset=['Potencia_Ativa'])
ds6['Potencia_Ativa'] = pd.to_numeric(ds6['Potencia_Ativa'], errors='coerce')
ds6 = ds6.dropna(subset=['Potencia_Ativa'])
print(f'Registros após limpeza: {len(ds6)}')
```

```python
# 4. Limiar 75% — alta potência ativa
max_pa  = ds6['Potencia_Ativa'].max()
lim_pa  = 0.75 * max_pa
alta6   = ds6[ds6['Potencia_Ativa'] > lim_pa]
qtd6    = len(alta6)
pct6    = (qtd6 / len(ds6)) * 100
print(f'Maior potência ativa: {max_pa:.3f} kW')
print(f'Limiar (75%): {lim_pa:.3f} kW')
print(f'Registros acima: {qtd6} ({pct6:.2f}%)')
```

```python
# 5. Segundo critério: alta potência ativa E climatização ligada
media_clim = ds6['Climatizacao'].mean()
alta6b = ds6[
    (ds6['Potencia_Ativa'] > lim_pa) &
    (ds6['Climatizacao'] > media_clim)
]
print(f'Climatização média: {media_clim:.2f} Wh')
print(f'Alta potência + climatização: {len(alta6b)} registros ({len(alta6b)/len(ds6)*100:.2f}%)')
print(f'Redução: {qtd6 - len(alta6b)} registros.')
```

```python
# Gráfico — Distribuição da potência ativa
plt.figure(figsize=(8, 4))
plt.hist(ds6['Potencia_Ativa'], bins=40, color='steelblue', edgecolor='white')
plt.axvline(lim_pa, color='red', linestyle='--', label=f'Limiar 75% ({lim_pa:.2f} kW)')
plt.title('Dataset 6 — Distribuição da Potência Ativa Residencial')
plt.xlabel('Potência Ativa (kW)')
plt.ylabel('Frequência')
plt.legend()
plt.tight_layout()
plt.savefig('ds6_grafico.png', dpi=100)
plt.close()
print('Gráfico salvo: ds6_grafico.png')
```
